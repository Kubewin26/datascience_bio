```python
def calculate_bmi(weight_kg, height):
    bmi = weight_kg / (height ** 2)
    rounded_bmi = round(bmi, 2)
    return rounded_bmi

my_bmi = calculate_bmi(75, 1.75)
print(f"My BMI is: {my_bmi}")
```

```python
def triage_glucose(glucose):
    if glucose < 70:
        return "Hypoglycemia"
    elif 70 <= glucose <= 99:
        return "Normal"
    elif 100 <= glucose <= 125:
        return "Prediabetes"
    else:
        return "Diabetes"

triage_glucose(105)
triage_glucose(126)

print(triage_glucose(105))
print(triage_glucose(126))
```

```python
def describe_patient(patient_id, status="Stable"):
    return f"Patient {patient_id} is currently {status}."

describe_patient("PT-101")
describe_patient("PT-202", "Critical") 

print(describe_patient("PT-101"))
print(describe_patient("PT-202", "Critical"))
```

```python
def log_biomarker(patient_id, marker_name, expression_level):
    return f"Patient {patient_id} has {marker_name} expression of {expression_level} RPKM."

print(log_biomarker("PT-01", "BRCA1", 14.8))
print(log_biomarker(expression_level=22.4, patient_id="PT-02", marker_name="TP53"))
```

```python
def calculate_average_reading(*readings):
    if len(readings) == 0:
        return 0.0
    average = sum(readings) / len(readings)
    return round(average, 2)

print(calculate_average_reading(120, 115, 125))
print(calculate_average_reading(98.6, 99.1, 98.4, 100.2, 98.9, 99.5))
```

```python
def countdown(n):
    if n == 0:
        print("Assay incubation complete!")
        return
    print(f"Time remaining: {n} seconds...")
    countdown(n - 1)

countdown(3)
```

```python
class Student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade  

    def introduce(self):
        print(f"My name is {self.name}, I am {self.age} years old, and I am in grade {self.grade}.")

student1 = Student("Ada", 15, 10)
student2 = Student("John", 16, 11)

student1.introduce()
student2.introduce()
```

```python
# Drill 2: Rectangle Class
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculateArea(self):
        return self.length * self.width

# Create two rectangle objects
rect1 = Rectangle(5, 3)
rect2 = Rectangle(10, 4)

# Display areas
print(f"A rectangle with length {rect1.length} and width {rect1.width} should have an area of {rect1.calculateArea()}.")
print(f"A rectangle with length {rect2.length} and width {rect2.width} should have an area of {rect2.calculateArea()}.")
```

```python
class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color
        self.speed = 0

    def accelerate(self):
        self.speed += 10

    def showDetails(self):
        print(f"Brand: {self.brand}, Color: {self.color}, Current Speed: {self.speed} km/h")

car1 = Car("Toyota", "Red")
car2 = Car("Honda", "Blue")

car1.accelerate()
car1.accelerate()

print(f"The first car's speed should be {car1.speed} km/h, and the second car's speed should still be {car2.speed} km/h.")

car1.showDetails()
car2.showDetails()
```

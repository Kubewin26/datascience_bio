```python
def calculate_bmi(weight_kg, height):
    bmi = weight_kg / (height ** 2)
    rounded_bmi = round(bmi, 2)
    return rounded_bmi

my_bmi = calculate_bmi(75, 1.75)
print(f"My BMI is: {my_bmi}")
```

```python
def triage_glucose(glucose):
    if glucose < 70:
        return "Hypoglycemia"
    elif 70 <= glucose <= 99:
        return "Normal"
    elif 100 <= glucose <= 125:
        return "Prediabetes"
    else:
        return "Diabetes"

triage_glucose(105)
triage_glucose(126)

print(triage_glucose(105))
print(triage_glucose(126))
```

```python
def describe_patient(patient_id, status="Stable"):
    return f"Patient {patient_id} is currently {status}."

describe_patient("PT-101")
describe_patient("PT-202", "Critical") 

print(describe_patient("PT-101"))
print(describe_patient("PT-202", "Critical"))
```

```python
def log_biomarker(patient_id, marker_name, expression_level):
    return f"Patient {patient_id} has {marker_name} expression of {expression_level} RPKM."

print(log_biomarker("PT-01", "BRCA1", 14.8))
print(log_biomarker(expression_level=22.4, patient_id="PT-02", marker_name="TP53"))
```

```python
def calculate_average_reading(*readings):
    if len(readings) == 0:
        return 0.0
    average = sum(readings) / len(readings)
    return round(average, 2)

print(calculate_average_reading(120, 115, 125))
print(calculate_average_reading(98.6, 99.1, 98.4, 100.2, 98.9, 99.5))
```

```python
def countdown(n):
    if n == 0:
        print("Assay incubation complete!")
        return
    print(f"Time remaining: {n} seconds...")
    countdown(n - 1)

countdown(3)
```

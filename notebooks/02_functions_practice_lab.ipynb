```python
def calculate_bmi(weight_kg, height):
    bmi = weight_kg / (height ** 2)
    rounded_bmi = round(bmi, 2)
    return rounded_bmi

my_bmi = calculate_bmi(75, 1.75)
print(f"My BMI is: {my_bmi}")
```

```python
def triage_glucose(glucose):
    if glucose < 70:
        return "Hypoglycemia"
    elif 70 <= glucose <= 99:
        return "Normal"
    elif 100 <= glucose <= 125:
        return "Prediabetes"
    else:
        return "Diabetes"

triage_glucose(105)
triage_glucose(126)

print(triage_glucose(105))
print(triage_glucose(126))
```

```python
def calculate_average_reading(*readings):
    if len(readings) == 0:
        return 0.0
    average = sum(readings) / len(readings)
    return round(average, 2)

print(calculate_average_reading(120, 115, 125))
print(calculate_average_reading(98.6, 99.1, 98.4, 100.2, 98.9, 99.5))
```

```python
# Drill 1: Encapsulation - The Secure Employee Class
class Employee:
    # Private class variable (The company baseline)
    _base_salary = 50000

    def __init__(self, name, level, position):
        self.name = name
        self.level = level
        self.position = position

    # Private calculation method (The silent accountant)
    def _calculate_salary(self):
        return Employee._base_salary + (Employee._base_salary * (self.level / 10))

    # Public method (The printed payslip)
    def generate_payslip(self):
        print(f"Name: {self.name} \nPosition: {self.position} \nLevel: {self.level} \nSalary: {self._calculate_salary()}")

    # Setter method (The safe front door to update base salary)
    def setBaseSalary(self, new_base_salary):
        Employee._base_salary = new_base_salary

# Test Employee 1
emp1 = Employee("Angela", 10, "Clerk")
emp1.generate_payslip()

# Update base salary via the official setter method
emp1.setBaseSalary(70000)

# Test Employee 2 under the new base salary
emp2 = Employee("Angela", 4, "Clerk")
emp2.generate_payslip()
```

```python
# Drill 2: Inheritance - The Universal Person & Specialized Roles

# 1. Base (Parent) Class
class Person:
    def __init__(self, name, age, gender, ID):
        self.name = name
        self.age = age
        self.gender = gender
        self.ID = ID

    def ID_card(self):
        print(f"Identity Number: {self.ID} \nName: {self.name} \nGender: {self.gender} \nAge: {self.age}")

# 2. Derived (Child) Class 1: Employee
class Employee(Person):
    def __init__(self, name, age, gender, ID, company, portfolio, salary):
        # Let the parent form fill in the standard identity boxes!
        Person.__init__(self, name, age, gender, ID)
        self.company = company
        self.portfolio = portfolio
        self.salary = salary

    def set_salary(self, new_salary):
        self.salary = new_salary

# 3. Derived (Child) Class 2: Student
class Student(Person):
    def __init__(self, name, age, gender, ID, school, course, level):
        Person.__init__(self, name, age, gender, ID)
        self.school = school
        self.course = course
        self.level = level

    def update_level(self):
        self.level += 100

# 4. Verification in action
emp = Employee("Ola", 45, "Male", "214892", "Rain", "Data Scientist", 20000)
emp.ID_card()  # Notice: Employee NEVER defined ID_card, it inherited it!

print("\n--- Student Details ---")
stu = Student("Moji", 34, "Female", "213456", "Rain", "AI", 200)
stu.ID_card()  # Inherited from Person!
stu.update_level()
print("Updated Student Level:", stu.level)
```

```python
# Drill 1: Abstraction with ABC and abstractmethod
from abc import ABC, abstractmethod

# 1. Abstract Base Class (The Hospital Rulebook)
class Staff(ABC):
    def __init__(self, name, job_title, staff_id):
        self.name = name
        self.job_title = job_title
        self.staff_id = staff_id

    # Concrete Method (Shared by all staff)
    def ID_Card(self):
        print(f"Staff ID: {self.staff_id} \nName: {self.name} \nPosition: {self.job_title}")

    # Abstract Method (The mandatory requirement!)
    @abstractmethod
    def calculate_payment(self):
        pass

# 2. Concrete Child 1: Part-Time Staff (Paid hourly)
class PT_Staff(Staff):
    def __init__(self, name, job_title, staff_id, hours_of_work):
        super().__init__(name, job_title, staff_id)
        self.hours_of_work = hours_of_work

    def calculate_payment(self):
        hourly_rate = 500
        amount = self.hours_of_work * hourly_rate
        return f"{amount:.2f}"

# 3. Concrete Child 2: Full-Time Staff (Paid by level formula)
class FT_Staff(Staff):
    def __init__(self, name, job_title, staff_id, level):
        super().__init__(name, job_title, staff_id)
        self.level = level

    def calculate_payment(self):
        base_pay = 50000
        amount = ((self.level / 10) * base_pay) + base_pay
        return f"{amount:.2f}"

# 4. Verify the Contract in Action
pt_employee = PT_Staff("Joshua", "Data Scientist", "111", 8)
ft_employee = FT_Staff("Joshua", "Data Scientist", "111", 8)

print("Part-Time Payment: #", pt_employee.calculate_payment())
print("Full-Time Payment: #", ft_employee.calculate_payment())
```

```python
# Drill 2: Polymorphism - Method Overriding

# 1. Base Class: Commercial Bank (Interest-based)
class Bank:
    def __init__(self, bank_name, establishment_year):
        self.bank_name = bank_name
        self.establishment_year = establishment_year

    def loan_repayment(self, amount, number_of_years, interest_rate):
        repayment = amount + (amount * (interest_rate / 100) * number_of_years)
        return f"{repayment:.2f}"

# 2. Subclass: Non-Interest Bank (Service-charge based)
class NON_Interest_Bank(Bank):
    # Overriding the parent's loan_repayment method
    def loan_repayment(self, amount, number_of_years, service_charge):
        repayment = amount + (service_charge * number_of_years)
        return f"{repayment:.2f}"

# 3. Testing Polymorphism
new_uba = Bank("UBA", 1906)
new_jaiz = NON_Interest_Bank("Jaiz", 2000)

print(f"Commercial Bank Repayment: #{new_uba.loan_repayment(5000000.5504, 5, 4.2)}")
print(f"Non-Interest Bank Repayment: #{new_jaiz.loan_repayment(500000, 3, 5000)}")
```

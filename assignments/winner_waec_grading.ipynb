```python
# ==============================================================================
# ASSIGNMENT 1: WAEC GRADING SYSTEM
# Course: Data Science & Machine Learning (Incubator Hub - DSA 2.0)
# Student: Winner Kubeyinje
# Task: Check the WAEC grade of a student from F9 up to A1
# ==============================================================================

# Step 1: Assign the student's test score
score = 80

# Step 2: Cascading conditional evaluation (Highest to Lowest)
if score < 0 or score > 100:
    grade = "INVALID"
    remark = "Score out of range (must be 0 - 100)"

elif score >= 75:
    grade = "A1"
    remark = "Excellent"

elif score >= 70:
    grade = "B2"
    remark = "Very Good"

elif score >= 65:
    grade = "B3"
    remark = "Good"

elif score >= 60:
    grade = "C4"
    remark = "Credit"

elif score >= 55:
    grade = "C5"
    remark = "Credit"

elif score >= 50:
    grade = "C6"
    remark = "Credit"

elif score >= 45:
    grade = "D7"
    remark = "Pass"

elif score >= 40:
    grade = "E8"
    remark = "Pass"

else:
    grade = "F9"
    remark = "Fail"

# Step 3: Output results using standard comma concatenation
print("Student Score:", score)
print("WAEC Grade:", grade)
print("Remark:", remark)
```

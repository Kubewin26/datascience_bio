## Why this matters
ML systems affect real people: hiring, credit, healthcare, policing, education.
A high accuracy score can still hide harm to a subgroup.

### Common bias sources
1. **Historical bias** — past unfair decisions are in the labels
2. **Representation bias** — some groups under-sampled
3. **Measurement bias** — proxy features (zip code ≈ race/income)
4. **Evaluation bias** — overall metrics hide subgroup failure


```python
import pandas as pd
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix

# Synthetic example: one group is under-represented and harder for the model
rng = np.random.default_rng(0)
X, y = make_classification(n_samples=2000, n_features=8, weights=[0.7, 0.3], random_state=42)
group = rng.choice(["A", "B"], size=len(y), p=[0.8, 0.2])

X_train, X_test, y_train, y_test, g_train, g_test = train_test_split(
    X, y, group, test_size=0.25, random_state=42, stratify=y
)
model = LogisticRegression(max_iter=2000).fit(X_train, y_train)
pred = model.predict(X_test)
print("Overall accuracy:", round(accuracy_score(y_test, pred), 3))

rows = []
for g in ["A", "B"]:
    mask = g_test == g
    acc = accuracy_score(y_test[mask], pred[mask])
    cm = confusion_matrix(y_test[mask], pred[mask])
    rows.append({"group": g, "n": int(mask.sum()), "accuracy": round(acc, 3)})
print(pd.DataFrame(rows))
```

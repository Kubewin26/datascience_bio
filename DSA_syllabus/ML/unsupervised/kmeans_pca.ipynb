# Unsupervised Learning — K-Means & PCA

**When:** Week 14 · Session A  
**Goal:** Find structure when there is **no label** to predict.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris, make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
```

## 1. K-Means intuition
- Choose **K** groups.
- Algorithm places K centers and assigns each point to the nearest center.
- Useful for customer segments, behavior groups, anomaly clues.


```python
X, y_true = make_blobs(n_samples=300, centers=3, cluster_std=1.1, random_state=42)
X = StandardScaler().fit_transform(X)

inertias, silhouettes = [], []
Ks = range(2, 8)
for k in Ks:
    km = KMeans(n_clusters=k, n_init=10, random_state=42)
    labels = km.fit_predict(X)
    inertias.append(km.inertia_)
    silhouettes.append(silhouette_score(X, labels))

fig, axes = plt.subplots(1, 2, figsize=(10, 3))
axes[0].plot(list(Ks), inertias, marker="o")
axes[0].set_title("Elbow (inertia)")
axes[0].set_xlabel("K")
axes[1].plot(list(Ks), silhouettes, marker="o")
axes[1].set_title("Silhouette score")
axes[1].set_xlabel("K")
plt.tight_layout()
plt.show()
```

```python
km = KMeans(n_clusters=3, n_init=10, random_state=42)
labels = km.fit_predict(X)
plt.figure(figsize=(5, 4))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap="viridis", s=25)
plt.scatter(km.cluster_centers_[:, 0], km.cluster_centers_[:, 1], c="red", s=120, marker="X")
plt.title("K-Means clusters (K=3)")
plt.show()
print("Silhouette:", round(silhouette_score(X, labels), 3))
```

## 2. PCA intuition
- PCA compresses many numeric features into a few components that keep most variance.
- Great for **visualization** and sometimes as a preprocessing step.
- Components are combinations of original features — interpret carefully.


```python
iris = load_iris()
Xi = StandardScaler().fit_transform(iris.data)
pca = PCA(n_components=2)
Xi_2d = pca.fit_transform(Xi)
print("Explained variance ratio:", np.round(pca.explained_variance_ratio_, 3))
print("Total variance kept:", round(pca.explained_variance_ratio_.sum(), 3))

plt.figure(figsize=(5, 4))
plt.scatter(Xi_2d[:, 0], Xi_2d[:, 1], c=iris.target, cmap="viridis", s=30)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Iris in 2D via PCA")
plt.show()
```

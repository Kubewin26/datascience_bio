## FUNCTIONS

### Defining a function

```python
def is_even(number):
    if (number % 2 == 0):
        print(number, "is an even number")
    else:
        print(number, "is an odd number")
```

### Calling a Function

```python
is_even(8)
```

```python
is_even(7.3)
```

### The return Keyword

```python
def is_multipleof3(number):
    if (number % 3 == 0):
        return True
    else:
        return False
```

```python
is_multipleof3(9)
```

```python
result = is_multipleof3(7)
```

```python
result
```

```python
if result:
    print("Number is a multiple of 3")
else:
    print("Number is not a multiple of 3")
```

### Default Parameter Values

```python
def person(name, age, nationality = "Nigerian"):
    print("My name is", name)
    print("I am", age, "years old")
    print("I am a", nationality)
```

```python
person("Amanda", 20)
```

```python
person("Brian", 17, "British")
```

### Keyword Arguments

```python
person(nationality ="Ghananian", name="Koffi", age= 20)
```

### A function that prints prime numbers between 1 and 50

```python
def is_prime(n):
    for factor in range(2, n):
        if(n % factor == 0):
            return False
    return True
```

```python
is_prime(5)
```

```python
is_prime(10)
```

```python
def print_prime_numbers():
    for number in range (1,50):
        if is_prime(number):
            print(number)
```

```python
print_prime_numbers()
```

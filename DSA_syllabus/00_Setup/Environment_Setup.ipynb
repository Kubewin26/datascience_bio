## 3. How a notebook works

- **Markdown cells** = notes and titles (like this one)
- **Code cells** = Python you can run
- Run a cell with **Shift + Enter**
- If you see `In [*]` the cell is still running — wait

### Common beginner errors

| Error | Meaning | Fix |
|-------|---------|-----|
| `NameError` | You used a name not defined yet | Run earlier cells first |
| `ModuleNotFoundError` | Package not installed | Install with conda/pip (below) |
| `FileNotFoundError` | Wrong file path | Check folder + spelling |
| `SyntaxError` | Typo in code | Check parentheses, quotes, colons |

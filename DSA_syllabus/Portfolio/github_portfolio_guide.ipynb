# GitHub Portfolio Guide (Beginner)

**When:** introduce lightly in Month 2; require for capstone  
**Goal:** Every student leaves with a public proof-of-work repo.


## Minimum portfolio setup
1. Create a free account at https://github.com
2. Install GitHub Desktop (easiest for beginners) **or** use git CLI
3. Create repo: `ai-ml-bootcamp-portfolio`
4. Suggested folders:

```
ai-ml-bootcamp-portfolio/
  01-python/
  02-pandas/
  03-visualization/
  04-ml/
  capstone/
  README.md
```

5. Push at least: 2 homework notebooks + capstone + README


## README template

```markdown
# AI/ML Bootcamp Portfolio — <Your Name>

## About
4-month beginner AI/ML program covering Python, Pandas, visualization, and supervised ML.

## Projects
1. **Pandas analysis** — short description + link
2. **EDA dashboard** — short description + link
3. **Capstone** — problem, approach, metric, demo screenshot

## Skills
Python · NumPy · Pandas · Matplotlib/Seaborn · scikit-learn · Streamlit

## How to run
```bash
pip install -r requirements.txt
jupyter notebook
```
```


## Commit message habit
- `add pandas salaries exercise`
- `fix missing-value handling in titanic cleaning`
- `complete capstone model comparison`

Avoid huge vague messages like `update` or `final final 2`.
